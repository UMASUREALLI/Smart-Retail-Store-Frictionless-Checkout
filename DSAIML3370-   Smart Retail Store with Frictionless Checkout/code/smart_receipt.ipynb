```python
from ultralytics import YOLO

model = YOLO("yolo11n.pt")

model.train(
    data="/content/Supermarket Product.v1i.yolov11/data.yaml",
    epochs=10,
    imgsz=640,
    batch=16
)
```

```python
from ultralytics import YOLO

model = YOLO("runs/detect/train/weights/best.pt")
```

```python
results = model.predict(
    source="/content/Supermarket Product.v1i.yolov11/test/images/20250717_211046_jpg.rf.f085ae45cb725eb2263d3b61dcb6077c.jpg",
    conf=0.5,
    save=True
)
```

```python
detected_items = []

for r in results:
    for box in r.boxes:
        class_id = int(box.cls[0])
        class_name = model.names[class_id]
        detected_items.append(class_name)

print("Detected Items:", detected_items)
```

```python
from collections import Counter

item_counts = Counter(detected_items)
print("Item Counts:", item_counts)
```

```python
class_names = list(model.names.values())
print(class_names)
```

```python
import pandas as pd

# Create inventory automatically
inventory = pd.DataFrame({
    "product_name": class_names,
    "price": [20 + i*10 for i in range(len(class_names))]  # auto-generate prices
})

inventory.to_csv("product_inventory.csv", index=False)

print(inventory)
```

```python
inventory_df = pd.read_csv("product_inventory.csv")
```

```python
results = model.predict(
    source="/content/Supermarket Product.v1i.yolov11/test/images",
    conf=0.5
)
```

```python
detected_items = []

for r in results:
    for box in r.boxes:
        class_id = int(box.cls[0])
        class_name = model.names[class_id]
        detected_items.append(class_name)
```

```python
from collections import Counter
item_counts = Counter(detected_items)
```

```python
from datetime import datetime
import pytz
import random
```

```python
ist = pytz.timezone("Asia/Kolkata")
now = datetime.now(ist)

current_date = now.strftime("%d-%m-%Y")
current_time = now.strftime("%H:%M:%S")
```

```python
transaction_id = "TXN" + str(random.randint(100000, 999999))
```

```python
total = 0

print("\n🧾 SMART RETAIL STORE")
print("================================")
print(f"Transaction ID : {transaction_id}")
print(f"Date           : {current_date}")
print(f"Time           : {current_time}")
print("================================")
print("Items Purchased:")
print("--------------------------------")

for item, count in item_counts.items():
    price = inventory_df.loc[
        inventory_df["product_name"] == item, "price"
    ].values[0]

    item_total = price * count
    total += item_total

    print(f"{item} x {count} = ₹{item_total}")

print("--------------------------------")
print(f"TOTAL AMOUNT = ₹{total}")
print("================================")
print("Thank You For Shopping 🛒")
```
